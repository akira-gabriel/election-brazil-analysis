```python
from pathlib import Path
import zipfile
import requests
import pandas as pd
import os
```

```python
url = "https://cdn.tse.jus.br/estatistica/sead/odsele/votacao_secao/votacao_secao_2022_BR.zip"

destiny_path = Path("..", "data", "raw")
destiny_path.mkdir(parents=True, exist_ok=True)
zip_path = destiny_path / url.split("/")[-1] 

r = requests.get(url, stream=True)

if r.status_code == 200:
    with open(zip_path, "wb") as f:
        for chunk in r.iter_content(chunk_size=1024 * 1024):
            f.write(chunk)
    print("sucess")
else:
    print(f"error: {r.status_code}")
```

```python
with zipfile.ZipFile(zip_path) as z:
    z.extractall(destiny_path)

origin_path = destiny_path / "votacao_secao_2022_BR.csv"
df = pd.read_csv(origin_path, sep=";", encoding="latin-1", dtype=str)
```

```python
os.remove(zip_path)
```
